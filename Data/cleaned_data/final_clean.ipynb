```python
import pandas as pd

df_aware = pd.read_csv('../aware/aware_cleaned.csv')
df_cbus = pd.read_csv('../cbus/cbus_cleaned.csv')
df_ngs = pd.read_csv('../ngs/ngs_cleaned.csv')
df_rest = pd.read_csv('../rest/rest_cleaned.csv')
```

```python
df_aware = df_aware.rename(columns={'Weighting (%)':'Weighting'})
df_cbus = df_cbus.rename(columns={'Weighting (%)':'Weighting'})
df_ngs = df_ngs.rename(columns={'Weighting (%)':'Weighting'})
df_rest = df_rest.rename(columns={'Weighting (%)':'Weighting'})
```

```python
# Filter only "Fixed income (PRIVATE DEBT)" rows
subset = df_cbus[df_cbus['Asset Class Name'] == "Fixed income (PRIVATE DEBT)"]

# Calculate subtotal row
subtotal = {
    "Asset Class Name": "Fixed income (PRIVATE DEBT)",
    "Name/Kind of Investment Item": "SUB TOTAL",
    "Int/Ext": "0",
    "Value (AUD)": subset["Value (AUD)"].sum(),
    "Weighting": subset["Weighting"].sum()
}

# Convert to DataFrame and append
df_cbus = pd.concat([df_cbus, pd.DataFrame([subtotal])], ignore_index=True)
```

```python
df_aware.to_csv('aware_final.csv', index=False)
df_cbus.to_csv('cbus_final.csv', index=False)
df_ngs.to_csv('ngs_final.csv', index=False)
df_rest.to_csv('rest_final.csv', index=False)
```
