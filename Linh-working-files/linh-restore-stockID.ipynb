```python
#filled NA in Listed Country column
#pay attention because sometimes "NA" is treated as empty cell then it will be removed. 

import pandas as pd
from pathlib import Path

# === Path ===
input_path = Path(r"D:\LinhDao\Programming\SUPERFUNdProject\final_data\aware_final.csv")

# --- Load ---
df = pd.read_csv(input_path)

# --- Ensure required columns ---
for col in ("Stock ID", "Listed Country"):
    if col not in df.columns:
        raise ValueError("File must contain 'Stock ID' and 'Listed Country' columns.")

# --- Conditions ---
mask_stockid_has_value = df["Stock ID"].notna() & df["Stock ID"].astype(str).str.strip().ne("")
mask_country_blank = df["Listed Country"].isna() | (df["Listed Country"].astype(str).str.strip() == "")

mask_to_fill = mask_stockid_has_value & mask_country_blank

# --- Fill ---
df.loc[mask_to_fill, "Listed Country"] = "NA"

# --- Save back; keep other blanks as empty cells (not 'NaN') ---
df.to_csv(input_path, index=False, na_rep="")

# --- Summary ---
print(f"✅ Updated file in place: {input_path}")
print(f"Rows where 'Listed Country' was filled with 'NA': {int(mask_to_fill.sum())}")
```

```python
#merge 2 columns listed country with stock ID to restore the original stock ID
import pandas as pd
from pathlib import Path

# === Path ===
input_path = Path(r"D:\LinhDao\Programming\SUPERFUNdProject\final_data\aware_final.csv")

# --- Load ---
# 👇 keep_default_na=False makes sure "NA" stays as a string, not NaN
df = pd.read_csv(input_path, keep_default_na=False)

# --- Ensure required columns ---
for col in ("Stock ID", "Listed Country"):
    if col not in df.columns:
        raise ValueError("File must contain 'Stock ID' and 'Listed Country' columns.")

# --- Merge into Stock ID only (keep Listed Country as is) ---
df["Stock ID"] = (
    df["Stock ID"].astype(str).str.strip() + " " + df["Listed Country"].astype(str).str.strip()
)

# --- Clean up ---
# 1) Strip trailing spaces
df["Stock ID"] = df["Stock ID"].str.strip()

# 2) Replace "nan", "NaN", "nan nan", etc. with empty string
df["Stock ID"] = df["Stock ID"].replace(
    to_replace=[r"^nan$", r"^NaN$", r"^nan nan$", r"^NaN NaN$"],
    value="",
    regex=True
)

# --- Save back ---
df.to_csv(input_path, index=False, na_rep="")

# --- Summary ---
print(f"✅ Updated file in place: {input_path}")

```
